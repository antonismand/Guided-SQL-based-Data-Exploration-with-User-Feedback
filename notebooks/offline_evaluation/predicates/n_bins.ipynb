```python
%load_ext autoreload
%autoreload 2

from sdss_eval.evaluator import create_tests, run_n_tests
from sdss_eval.sqlparser import schema
from sdss_eval.discretize import Binning
from sdss_eval.evaluator import ReturnAll, evaluate_conditions
import pandas as pd

n_bins_to_test = [10, 30, 50, 100, 300, 500]

cols = [
            "specobj.z",
            "specobj.ra",
            "specobj.dec",
            "photoobj.ra",
            "photoobj.g",
            "photoobj.dec",
            "photoobj.u",
        ]
```

```python
for col in cols:
    print("---------- Tuning", col, "----------")
    df = pd.DataFrame(columns=["strategy", "n_bins", "score"])
    for strategy in ["uniform", "quantile", "kmeans"]:
        for n_bins in n_bins_to_test:
        
            Binning([col]).create_bins([n_bins], strategy=strategy)


            db = schema()
            db.create_bins(discretization_method=strategy, columns = [col])

            scr = evaluate_conditions(ReturnAll(), db)
            df = pd.concat([pd.DataFrame([[strategy, n_bins, scr[-1]]], columns=df.columns), df], ignore_index=True)
        

        sorted = df.loc[df['strategy'] == strategy].sort_values("score", ascending=False)
        print(sorted)
        best_n_bins = sorted['n_bins'].iloc[0]
        print("Best n_bins is",best_n_bins)
        Binning([col]).create_bins([best_n_bins], strategy=strategy)

```

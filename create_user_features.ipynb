## Manually create other log

```python
from sqlcompletions.processing import user_context

# SELECT Clause: n_components:  8, 14, 20
# FROM/WHERE Clause: n_components: 6, 10, 14
for clause in [0,1,2]:
    features = user_context(clause=clause, log_file="log_th500.json")
    features.PCA_fit(n_components=14)
    features.transform(add_one=True)
```

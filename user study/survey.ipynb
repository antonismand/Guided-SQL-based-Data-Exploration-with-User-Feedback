```python
import pandas as pd
df = pd.read_csv("survey.csv", delimiter=",")
df.columns
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import tikzplotlib

def tikzplotlib_fix_ncols(obj):
    """
    workaround for matplotlib 3.6 renamed legend's _ncol to _ncols, which breaks tikzplotlib
    """
    if hasattr(obj, "_ncols"):
        obj._ncol = obj._ncols
    for child in obj.get_children():
        tikzplotlib_fix_ncols(child)


ax = sns.countplot(x="value", hue="variable", 
              order=["Not at all familiar", "Slightly familiar","Somewhat familiar","Moderately familiar","Extremely familiar"], 
              data=pd.melt(df[['familiarity_1','familiarity_2']])
              )
ax.set(ylabel="Number of votes", xlabel="")
ax.set_xticklabels(labels=["Not at all\n familiar", "Slightly\nfamiliar","Somewhat\nfamiliar","Moderately\nfamiliar","Extremely\namiliar"])
leg = plt.legend(title="", labels=["SQL","Cordis"])



tikzplotlib_fix_ncols(leg)

tikzplotlib.save(
            "familiarity.tex",
            extra_axis_parameters=["scaled x ticks=false", "scaled y ticks=false"],
            axis_height="\\figH",
            axis_width="\\figW",
        )
```

```python
ax = sns.countplot(x="value", hue="variable", 
              order=["Strongly disagree", "Disagree","Neither agree nor disagree","Agree","Strongly agree"], 
              data=pd.melt(df[['Q17_1', 'Q17_2', 'Q17_3', 'Q17_4']])
              )
ax.set(ylabel="Number of votes", xlabel="")
leg = plt.legend(title="", loc='upper left' )
tikzplotlib_fix_ncols(leg)

tikzplotlib.save(
            "g1.tex",
            extra_axis_parameters=["scaled x ticks=false", "scaled y ticks=false"],
            axis_height="\\figH",
            axis_width="\\figW",
        )
```
